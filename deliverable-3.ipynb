### Task 1 - Character class

In this task, we implement the basic functionality of the Character class, which represents a character in the RPG simulator.

The class includes methods for attacking other characters, taking damage, and restoring health. The take_damage() and heal() methods ensure that HP stays within the valid range of 0 to hp_max.

We also implement the is_alive property, which returns whether the character has more than 0 HP remaining.

```python
# Task 1 - Character class demonstration

from character import Character

# Create two characters with different attributes
hero = Character("Aria", 100, 15)
enemy = Character("Goblin", 60, 8)

# Display the characters' attributes
print("Hero:")
print("Name:", hero.name)
print("HP:", hero.hp)
print("Max HP:", hero.hp_max)
print("Damage:", hero.damage)
print("Weapon:", hero.weapon)

print("\nEnemy:")
print("Name:", enemy.name)
print("HP:", enemy.hp)
print("Max HP:", enemy.hp_max)
print("Damage:", enemy.damage)
print("Weapon:", enemy.weapon)
```

```python
# Demonstrate the attack method

print("Goblin's HP before attack:", enemy.hp)

hero.attack(enemy)

print("Goblin's HP after attack:", enemy.hp)
```

```python
# Demonstrate healing and HP limits

# Heal Goblin after the attack
enemy.heal(10)
print("HP after healing:", enemy.hp)

# Try healing beyond maximum HP
enemy.heal(100)
print("HP after excessive healing:", enemy.hp)

# Take more damage than remaining HP
enemy.take_damage(100)
print("HP after excessive damage:", enemy.hp)

# Check if Goblin is still alive
print("Is Goblin alive?", enemy.is_alive)
```

```python
# Demonstrate character introduction and string representations

hero.introduce()

print(str(hero))
print(repr(hero))
```

### Results and Discussion

The tests demonstrate that the `Character` class works as expected.

When Aria attacked Goblin, Goblin's HP decreased from 60 to 45, confirming that the `attack()` and `take_damage()` methods work together correctly.

The healing tests showed that HP increased from 45 to 55, but could not exceed the maximum of 60. Similarly, excessive damage reduced HP to 0 without allowing negative values. This demonstrates how `min()` and `max()` enforce the HP limits.

The `is_alive` property returned `False` when Goblin reached 0 HP. The `introduce()`, `__str__()` and `__repr__()` methods also produced the expected character descriptions.

Overall, the tests confirm that the basic character functionality works correctly and provides a foundation for the specialised character classes and battle system.

### Task 2 - Dice
In this task, we created a Dice class. The dice has six sides by default, but it is also possible to choose a diffrent number of sides.

The roll() method returns a random number between 1 and the number of sides. 

```python
from dice import Dice

d6 = Dice()
d8 = Dice(8)

print("Six-sided dice:", d6.roll())
print("Eight-sided dice:", d8.roll())
print("Calling the dice directly:", d6())

print("Number of sides (d6):", d6.sides)
print("Number of sides (d8):", d8.sides)

print("Dice representation:", repr(d6))
```

### Results and Discussion

The tests showed that both dice returned values within their expected ranges (1–6 and 1–8). The `__call__()` method also worked, allowing the dice to be rolled directly using `d6()`.

The `sides` attribute correctly displayed the number of sides for each dice, while `__repr__()` provided a readable representation of the object.

Overall, the results demonstrate that the Dice class works as expected in the tested examples.

## Task 3 – Weapon Class

### Overview and Task Description

In this task, we introduce the `Weapon` class, which uses a `Dice` object to calculate random damage through the `roll_damage()` method.

We also extend the `Character` class to support optional weapons. When a character has a weapon equipped, its attacks use the weapon's damage instead of the character's base damage.

```python
from dice import Dice
from items import Weapon

sword = Weapon("Sword", Dice(8))
axe = Weapon("Axe", Dice(12))

print("Weapon:", sword.name)
print("Damage dice:", sword.damage_dice)
print("Damage:", sword.roll_damage())

print("Weapon:", axe.name)
print("Damage dice:", axe.damage_dice)
print("Damage:", axe.roll_damage())
```

```python
# Task 3 - Character with and without a weapon

from character import Character

# Create characters
armed_hero = Character("Aria", damage=15, weapon=sword)
unarmed_hero = Character("Luna", damage=15)

# Create targets
target1 = Character("Goblin", hp_max=100)
target2 = Character("Orc", hp_max=100)

# Demonstrate attacks
armed_hero.attack(target1)
unarmed_hero.attack(target2)

print("Aria's weapon:", armed_hero.weapon.name)
print("Goblin HP after weapon attack:", target1.hp)

print("\nLuna's weapon:", unarmed_hero.weapon)
print("Orc HP after regular attack:", target2.hp)
```

### Results and Discussion

The tests showed that the `Weapon` class successfully uses dice to generate random damage. The Sword (Dice(8)) dealt 6 damage, while the Axe (Dice(12)) dealt 8 damage.

When equipped with a weapon, the `Character` class used the weapon's dice roll instead of its base damage. Without a weapon, the character dealt its default damage of 15.

Overall, the results demonstrate how weapons and dice work together through object composition, allowing characters to deal variable damage.

## Task 4 – Warrior and Mage

### Overview and Task Description

In this task, we extend the `Character` class with two subclasses: `Warrior` and `Mage`.

The `Warrior` class introduces an `armor` attribute and overrides `take_damage()` to reduce incoming damage.

The `Mage` class introduces `mana` and `mana_max`, and overrides `attack()` to deal double damage using mana. When mana is insufficient, the mage performs a regular attack and restores 5 mana.

Both subclasses inherit the basic functionality of `Character`.

```python
# Task 4 - Warrior demonstration

from character import Warrior

# Create a warrior with 100 HP, 15 damage and 8 armor
warrior = Warrior("Thor", hp_max=100, damage=15, armor=8)

# Display attributes
print("Name:", warrior.name)
print("HP:", warrior.hp)
print("Max HP:", warrior.hp_max)
print("Damage:", warrior.damage)
print("Armor:", warrior.armor)
print("Weapon:", warrior.weapon)

# Demonstrate armor protection
warrior.take_damage(30)
print("\nHP after taking 30 damage:", warrior.hp)

# Demonstrate that armor can block all damage
warrior.take_damage(5)
print("HP after taking 5 additional damage:", warrior.hp)
```

```python
# Task 4 - Mage demonstration

from character import Mage, Character

# Create a mage and an enemy
mage = Mage("Merlin", hp_max=100, damage=10)
target = Character("Goblin", hp_max=100)

# Display attributes
print("Name:", mage.name)
print("HP:", mage.hp)
print("Damage:", mage.damage)
print("Mana:", mage.mana)
print("Max mana:", mage.mana_max)
print("Weapon:", mage.weapon)

# Demonstrate magic attacks and mana consumption
for i in range(5):
    mage.attack(target)
    print(f"\nAttack {i + 1}:")
    print("Remaining mana:", mage.mana)
    print("Target HP:", target.hp)
```

```python
# Demonstrate Mage attacks with an equipped weapon

from dice import Dice
from items import Weapon

# Create a weapon and equip the mage
staff = Weapon("Magic Staff", Dice(6))

armed_mage = Mage("Gandalf", damage=10, mana=10, weapon=staff)
enemy_mage = Character("Orc", hp_max=100)

print("Weapon:", armed_mage.weapon.name)
print("Starting mana:", armed_mage.mana)

# First attack: magic (ignores weapon)
armed_mage.attack(enemy_mage)

print("\nAfter magic attack:")
print("Mana:", armed_mage.mana)
print("Enemy HP:", enemy_mage.hp)

# Second attack: regular attack (uses weapon)
armed_mage.attack(enemy_mage)

print("\nAfter regular attack:")
print("Mana:", armed_mage.mana)
print("Enemy HP:", enemy_mage.hp)
```

### Results and Discussion

The tests confirmed that both subclasses work as expected.

The Warrior's armor reduced incoming damage from 30 to 22, leaving 78 HP. An additional attack of 5 damage was completely blocked by the armor.

The Mage successfully used 10 mana to deal double damage (20 HP). When mana dropped below 10, regular attacks were performed and 5 mana was restored after each attack.

The weapon test also confirmed that magical attacks ignore equipped weapons, while regular attacks use weapon-based damage.

Overall, the results demonstrate how inheritance and method overriding allow different character specialisations.
